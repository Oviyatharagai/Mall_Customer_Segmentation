**Data Set**

```python
import pandas as pd


!pip install openpyxl

data=pd.read_excel('dataset.xlsx')
df=pd.DataFrame(data)

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

display(df)
```

**Data Segmentation**

```python
print("First 5 rows:")
print(df.head())

print("\nShape:")
print(df.shape)

print("\nColumn names:")
print(df.columns)
```

**Check Missing values**

```python
print(df.isnull().sum())
```

**Check Duplicates**

```python
print("Number of duplicates:", df.duplicated().sum())
```

**Check Datatypes**

```python
print(df.dtypes)
```

**Finding Categorical & Numerical Columns**

```python
numerical = df.select_dtypes(include='number').columns
categorical = df.select_dtypes(include='object').columns

print("Numerical columns:")
print(numerical)

print("\nCategorical columns:")
print(categorical)
```

Clean Missing Values

```python
df[numerical] = df[numerical].fillna(df[numerical].mean())
df['Age'] = df['Age'].fillna(df['Age'].mean())
df.isnull().sum()
```

```python
df['Profession'] = df['Profession'].fillna(df['Profession'].mode()[0])
print(df.isnull().sum())
```

**Bar Chart**

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

df['Profession'].value_counts().plot(kind='bar')

plt.xlabel('Profession') # Updated x-label to match the plotted column
plt.ylabel('Count')
plt.title('Distribution by Profession') # Updated title
plt.show()
```

**Pie Chart**

```python
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Gender Distribution')
plt.ylabel('')
plt.show()
```

**Scatter Plot**

```python
plt.scatter(df['Age'], df['Annual_Income_k$'])
plt.xlabel('Age')
plt.xlabel('Annual_Income_k$')
plt.title('Age vs Annual Income')
plt.show()
```
